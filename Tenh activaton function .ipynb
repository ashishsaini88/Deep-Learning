```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler 
from sklearn.metrics import accuracy_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense 
```

```python
df = pd.read_csv(r"C:\Users\KOIN\Downloads\titanic (1).csv")
df.head()
```

```python
df = df.drop(["Name" , "Ticket" , "Cabin" ], axis=1)
```

```python
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Embarked'].fillna('S', inplace=True)

df = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True)

df= df.dropna(subset=['Fare'])
```

```python
x = df.drop('Survived', axis=1)
y = df['Survived']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
model = Sequential()
```

```python
model.add(Dense(16 , input_dim = x_train_scaled.shape[1] , activation="tanh"))
```

```python
model.add(Dense(1 , activation="sigmoid"))
```

```python
model.compile(loss = "binary_crossentropy" , optimizer="adam" , metrics=["accuracy"])
```

```python
model.fit(x_train_scaled , y_train , epochs=40 , batch_size=10 , verbose=1)
```

```python
accuracy = model.evaluate(x_test_scaled , y_test)

print("the accuracy of model using tenh =" , accuracy )

y_pred = model.predict(x_test_scaled[:5])

print("pridicted probability using tanh activation into hidden layes :")
print(y_pred)
```
